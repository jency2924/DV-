```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel("/content/drive/MyDrive/AAPL.xlsx")
```

```python
df["Date"] = pd.to_datetime(df["Date"],utc=True)
df['Date'].dtype
```

```python
df = df.sort_values("Date")
```

```python
df["Date"] = pd.to_datetime(df["Date"],utc=True)
df
```

```python
plt.figure(figsize=(16, 8))

plt.plot(df.index, df["Open"], label="Open", color="orange")
plt.plot(df.index, df["High"], label="High", color="red")
plt.plot(df.index, df["Low"], label="Low", color="blue")
plt.plot(df.index, df["Close"], label="Close", color="green")

plt.title("AAPL OHLC Prices")
plt.xlabel("Date")
plt.ylabel("Prices")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(16, 6))

plt.plot(df.index, df["Volume"], label="Volume" , colour="green")

plt.title("AAPL Trading Volume")
plt.xlabel("Date")
plt.ylabel("Volume")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python

df["MA20"] = df["Close"].rolling(window=20).mean()

df["MA50"] = df["Close"].rolling(window=50).mean()
```

```python
plt.figure(figsize=(16, 8))

plt.plot(df.index, df["Close"], label="Daily Close" , color="orange")
plt.title("AAPL Closing Price with 20-Day and 50-Day Moving Averages")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python

plt.figure(figsize=(12, 6))

plt.plot(df['Date'], df["MA20"], label="20-Day MA")

plt.title("Appl Price Moving Averages for short term")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df['Date'], df["MA50"], label="50-Day MA")

plt.title("Appl Moving Averages for long term")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python

plt.figure(figsize=(12, 6))

plt.plot(df.index, df["Close"], label="Daily Close", color="red")
plt.plot(df.index, df["MA20"], label="20-Day MA",color="blue")
plt.plot(df.index, df["MA50"], label="50-Day MA" , color="black")

plt.title("Appl and Moving Averages")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python

df["Daily_Return"] = df["Close"].pct_change()


df
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    df["Daily_Return"].dropna(),
    bins=50,
    kde=True
)

plt.title("Appl Returns")
plt.xlabel("Daily Return")
plt.ylabel("Frequency")
plt.show()
```

```python

df["Rolling_Volatility_20"] = (
    df["Daily_Return"]
    .rolling(window=20)
    .std()
)
df["Rolling_Volatility_20"].dropna()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df.index,
    df["Rolling_Volatility_20"],
    color="green"
)

plt.title("Appl Rolling Volatility")
plt.xlabel("Date")
plt.ylabel("Volatility")
plt.xticks(rotation=45)
plt.show()
```
