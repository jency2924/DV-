**Correct format**

```python
data["Date"] = pd.to_datetime(data["Date"], errors="coerce")


print(data["Date"].head())

print("\nDate column data type:")
print(data["Date"].dtype)
```

**Analyze The Stock Price**

**Average Opening Price **

```python
average_opening_price = data["Open"].mean()

print("Average Opening Price: $", round(average_opening_price, 2))
```

**Average Closing Price **

```python
average_closing_price = data["Close"].mean()

print("Average Closing Price: $", round(average_closing_price, 2))
```

**Highest Record in High Column**

```python
highest_record = data.loc[data["High"].idxmax()]

print(highest_record)
```

**lowest record in  low column**

```python
lowest_record = data.loc[data["Low"].idxmin()]

print(lowest_record)
```

**Calculate Daily Data**

```python
data["Delta"] = data["Close"] - data["Open"]
```

** Find 10 days with the highest price change**

```python
top_10 = data.nlargest(10, "Delta")
```

```python

print("10 Days with Highest Price Change:")
print(top_10[["Date", "Open", "Close", "Delta"]])
```

**Calculate daily percentage return**

**Highest **

```python
data["Daily_Return"] = ((data["Close"] - data["Open"]) / data["Open"]) * 100

top_10_returns = data.nlargest(10, "Daily_Return")
print("10 Days with Highest Return:")
print(top_10_returns[["Date", "Open", "Close", "Daily_Return"]])
```

**Lowest**

```python
lowest_10 = data.nsmallest(10, "Daily_Return")

print("10 Days with Lowest Return:")
print(lowest_10[["Date", "Open", "Close", "Daily_Return"]])
```
